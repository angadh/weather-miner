```python
import datetime
import math 

def RaceTimes(hours=0, minutes=0, seconds=0):
    """
    Convert run time from minutes to hours, minutes and seconds.
    """
    return datetime.timedelta(hours=hours, minutes=minutes, seconds=seconds)


# We need overall marathon time (default = 26.2 miles)
marathonDistance = 26.2

# What's the desired finish time.
desiredTime = RaceTimes(hours=4, minutes=0)
print("Desired Marathon Time : ", desiredTime)

# Now compute the average speed needed to achieve this.
desiredPace = desiredTime / marathonDistance
print("Effective Marathon Pace : ", desiredPace)

# Run intervals in minutes
runMax = 6
runMin = 1

### Fixed walk interval of 30 seconds
walkInterval = RaceTimes(seconds=30)
print(walkInterval)

### Assume that the walking speed is 4 miles per hour or 15 minutes per mile speed
walkSpeed = RaceTimes(minutes=15)
```

```python

### Sample execution within the for loop
runInterval = RaceTimes(minutes=4)

numIntervals = desiredTime / (runInterval + walkInterval)

# Divide the intervals into full and partial intervals
(partialIntervals, fullIntervals) = math.modf(numIntervals)

# This is the total time spent running.
runTime = (fullIntervals + partialIntervals) * runInterval
print("Time spend running = ", runTime)

# This is the total time walking
walkTime = fullIntervals * walkInterval
print("Time spend walking = ", walkTime)

# Distance covered walking
walkDistance = walkTime / walkSpeed
print("Distance covered walking :", round(walkDistance, 2), "miles")

runDistance = marathonDistance - walkDistance
print("Distance covered running :", round(runDistance, 2), "miles")

runSpeed = runTime / runDistance
print("Required running speed (time per mile) :", runSpeed, "\n\n\n")
```

```python
for runInterval in [RaceTimes(minutes=j/10) for j in range(10*runMax, 10*runMin-1, -5)]:
    numIntervals = desiredTime / (runInterval + walkInterval)

    print("::::::::::: For a Run / Walk interval of ", runInterval, "/", walkInterval, ":::::::::::")

    # Divide the intervals into full and partial intervals
    (partialIntervals, fullIntervals) = math.modf(numIntervals)
    
    # This is the total time spent running.
    runTime = (fullIntervals + partialIntervals) * runInterval
    print("Time spend running = ", runTime)
    
    # This is the total time walking
    walkTime = fullIntervals * walkInterval
    print("Time spend walking = ", walkTime)
    
    # Distance covered walking
    walkDistance = walkTime / walkSpeed
    print("Distance covered walking :", round(walkDistance, 2), "miles")
    
    runDistance = marathonDistance - walkDistance
    print("Distance covered running :", round(runDistance, 2), "miles")
    
    runSpeed = runTime / runDistance
    print("Required running speed (time per mile) :", runSpeed, "\n\n\n")
```
