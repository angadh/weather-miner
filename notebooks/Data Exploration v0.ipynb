```python
import netCDF4
import os 

import numpy as np 
import tensorflow as tf 

reanalysis_path = "../data/reanalysis/"
reanalyis_files = os.listdir(reanalysis_path)
print(reanalyis_files)

# Training and Validation on data from 2023 and 2024.
# Testing on data from 2025
```

```python
reanalysis_vars = set([file.split(".")[0] for file in reanalyis_files])
reanalysis_years = set([file.split(".")[1] for file in reanalyis_files])
print(reanalysis_vars)
print(reanalysis_years)
```
